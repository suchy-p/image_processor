```python
def get_random_samples(image, sampling_range, sample_size):
    image = image
    height, width = np.shape(image)[:2]
    sampling_range = sampling_range
    sample_size = sample_size
    samples = []

    for h in range(height):
        for w in range(width):
            height_sample = np.random.randint(
                low= 0 if h == 0 else h - sampling_range,
                high= h + sampling_range if sampling_range >= h else h,
                size=sample_size,
                dtype=int
            )
            width_sample = np.random.randint(
                low= 0 if w == 0 else w - sampling_range,
                high= w + sampling_range if sampling_range >= w else w,
                size=sample_size,
                dtype=int
            )
            samples.append((height_sample, width_sample))

    return samples
```

```python
import numpy as np
def test_get_random_samples():
    test_array = np.empty([1400, 500], dtype=np.uint8)
    test_sampling_ranges = iter([200, 800, 1400])
    test_samples = [3, 10, 50]

    for samples in test_samples:
        assert get_random_samples(test_array,
                                       next(test_sampling_ranges),
                                       test_samples
                                       ) == len(range(samples))
        for test_sample in get_random_samples(test_array,
                                       test_sampling_ranges,
                                       test_samples):
            assert len(test_sample) == 2
```
